**(7)**

Briefly explain the practical meaning of the expectation and variance in this setting.


The Expectation of 0.75 serves as the "steady-state" baseline for the electricity district’s long-term planning. While no single day will ever result in exactly 0.75 failures, this figure represents the average demand on maintenance services over many months. For a manager, this suggests that the system typically experiences less than one emergency per day, allowing them to calculate the total volume of spare parts, fuel for service vehicles, and annual labor costs needed to keep the grid operational. It effectively converts the randomness of daily reports into a stable figure used for financial budgeting and general resource allocation.

The Variance of 0.7875 provides a critical measure of the system's volatility and the inherent risk of the grid's failure patterns. A high variance indicates that while "quiet days" with zero failures are frequent, the district must still account for the "spikiness" of the data—those rare but high-impact days where three transformers fail simultaneously. Practically, this value dictates the size of the emergency response team; the district cannot staff based on the average alone because the variance tells them that the workload will often swing significantly above that average. It essentially measures how much "buffer room" or emergency standby capacity is required to handle unpredictable surges in equipment failure

```python
import numpy as np

# Define the Random Variable values and their probabilities
x_values = np.array([0, 1, 2, 3])
probabilities = np.array([0.50, 0.30, 0.15, 0.05])

# 4. Compute Expectation E[X]
expectation = np.sum(x_values * probabilities)

# 5. Compute Variance Var(X)
# Formula: E[X^2] - (E[X])^2
expectation_x_squared = np.sum((x_values**2) * probabilities)
variance = expectation_x_squared - (expectation**2)

print(f"Expectation E[X]: {expectation}")
print(f"Variance Var(X): {variance}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
bars = plt.bar(x_values, probabilities, color='skyblue', edgecolor='navy', alpha=0.7)

# Adding labels and titles
plt.title('Probability Mass Function (PMF) of Transformer Failures', fontsize=14)
plt.xlabel('Number of Failures (X)', fontsize=12)
plt.ylabel('Probability P(X=x)', fontsize=12)
plt.xticks(x_values)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Add probability labels on top of bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.01, f'{yval:.2f}', ha='center', va='bottom')

# Highlight the Expectation (Mean)
plt.axvline(expectation, color='red', linestyle='dashed', linewidth=2, label=f'Mean (E[X] = {expectation})')
plt.legend()

plt.show()
```

```python
# Cumulative probabilities for question 6
p_ge_1 = np.sum(probabilities[x_values >= 1])
p_ge_2 = np.sum(probabilities[x_values >= 2])

print(f"P(X >= 1): {p_ge_1}")
print(f"P(X >= 2): {p_ge_2}")

# Plotting the Cumulative Distribution Function (CDF)
cdf = np.cumsum(probabilities)

plt.figure(figsize=(8, 5))
plt.step(x_values, cdf, where='post', color='darkorange', linewidth=2.5)
plt.title('Cumulative Distribution Function (CDF)', fontsize=14)
plt.xlabel('Number of Failures (x)', fontsize=12)
plt.ylabel('P(X <= x)', fontsize=12)
plt.xticks(x_values)
plt.grid(True, alpha=0.3)
plt.fill_between(x_values, cdf, step="post", alpha=0.2, color='orange')
plt.show()
```

1 .  Construct  a Probability  space (Ω,ℱ,P) Suitable for this model.


given distribution is a Beta distribution


$$f(x) = \begin{cases} \frac{x^1(1-x)^{17}}{B(2,18)}, & 0 < x < 1 \\ 0, & \text{otherwise} \end{cases}$$

So the Sample space :$$Ω = {[0,1]}$$

Then the σ - Algebra is :$$\mathcal{F} = \mathcal{B}([0, 1])$$


Probability Measure : $$P(A) = \int_{A} f(x) \, dx$$

$$P(A) = \int_{A} \frac{x(1-x)^{17}}{B(2, 18)} \, dx \quad \forall A \in \mathcal{F}$$

2. Define **X** as a continuous random variable on this space.



$$X: \Omega \to \mathbb{R}$$

Specifically, we define $X$ as the identity mapping, where for every outcome $\omega$ in our sample space $\Omega = [0, 1]$, the random variable takes that exact value:$$X(\omega) = \omega$$

---

Write down the probability density function of **X**



$$f(x) = \begin{cases} \frac{x^1(1-x)^{17}}{B(2,18)}, & 0 < x < 1 \\ 0, & \text{otherwise} \end{cases}$$

---

3. Verify that **f** is a valid probability density function.


$$I =\int_{0}^{1} \left( \frac{x(1-x)^{17}}{B(2, 18)} \right) \, dx $$

$$I = \frac{1}{B(2, 18)} \int_{0}^{1} x^{2-1}(1-x)^{18-1} \, dx$$


The integral part $\int_{0}^{1} x^{2-1}(1-x)^{18-1} \, dx$ is, by definition, $B(2, 18)$. Substituting the Gamma identities for both the numerator and

denominator:$$I = \frac{1}{\left( \frac{\Gamma(2)\Gamma(18)}{\Gamma(20)} \right)} \cdot \left( \frac{\Gamma(2)\Gamma(18)}{\Gamma(20)} \right)$$




$$\Gamma(2) = 1! = 1$$
$$\Gamma(18) = 17!$$
$$\Gamma(20) = 19!$$

Substituting these into the identity:$$B(2, 18) = \frac{1 \cdot 17!}{19!} = \frac{17!}{19 \cdot 18 \cdot 17!} = \frac{1}{342}$$Now, returning to the full expression:$$I = \frac{\int_{0}^{1} x(1-x)^{17} \, dx}{1/342}$$$$I = \frac{1/342}{1/342}$$

$$I = 1 $$


 $\int_{0}^{1} f(x) \, dx = 1$ and $f(x) \geq 0$ for all $x$, $f(x)$ is a valid probability density function.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Define the shape parameters based on your derivation
alpha = 2
beta_val = 18

# Generate 500 points between 0 and 1 for a smooth curve
x = np.linspace(0, 1, 500)

# Calculate the probability density function (PDF)
y = beta.pdf(x, alpha, beta_val)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, label=f'Beta({alpha}, {beta_val})', color='#2E86C1', lw=2.5)

# Formatting the graph
plt.title(r'Probability Density Function: $f(x) = \frac{x(1-x)^{17}}{B(2, 18)}$', fontsize=14)
plt.xlabel('$x$', fontsize=12)
plt.ylabel('$f(x)$', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)

# Highlight the mean (E[X] = 0.1)
plt.axvline(0.1, color='red', linestyle=':', label='Mean (0.1)')

plt.legend()
plt.tight_layout()

# Show the plot
plt.show()
```
