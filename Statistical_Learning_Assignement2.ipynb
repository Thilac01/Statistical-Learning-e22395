To define $X$ as a random variable on the space $(\Omega, \mathcal{F}, P)$, we define it as a function $X: \Omega \rightarrow \mathbb{R}$ such that for every outcome $\omega \in \Omega$:

$$X(\omega) = \omega$$


Explicitly, this mapping is:

* $X(0) = 0$
* $X(1) = 1$
* $X(2) = 2$
