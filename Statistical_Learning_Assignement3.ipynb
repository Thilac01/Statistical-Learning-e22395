#Q1) Prediction of fraudulent transactions

----

1) **Probability Space**

* **Sample Space:** $\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$
* **Sigma Algebra:** $\mathfrak{F} = \mathcal{P}(\Omega)$

$$\mathfrak{F} = \left\{ \emptyset, \{\omega_1\}, \{\omega_2\}, \{\omega_3\}, \{\omega_4\}, \{\omega_1,\omega_2\}, \{\omega_1,\omega_3\}, \{\omega_1,\omega_4\}, \{\omega_2,\omega_3\}, \{\omega_2,\omega_4\}, \{\omega_3,\omega_4\}, \{\omega_1,\omega_2,\omega_3\}, \{\omega_1,\omega_2,\omega_4\}, \{\omega_1,\omega_3,\omega_4\}, \{\omega_2,\omega_3,\omega_4\}, \Omega \right\}$$


* **Probability measure P**

$$P(\{\omega_1\}) = 0.50, \quad P(\{\omega_2\}) = 0.10, \quad P(\{\omega_3\}) = 0.20, \quad P(\{\omega_4\}) = 0.20$$

---

# 2)  $\sigma$-algebra Generated by $Y$, $\sigma(Y)$


 $Y$:$\{Y = 0\} = Y^{-1}(\{0\}) = \{\omega_1, \omega_2\}$$\{Y = 1\} = Y^{-1}(\{1\}) = \{\omega_3, \omega_4\}$

 Taking all unions, intersections, and complements of these foundational blocks yields:$$\sigma(Y) = \left\{ \emptyset, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}, \Omega \right\}$$

---


# 3)  Measurability of $E[X \mid Y]$


By definition, the conditional expectation $E[X \mid Y]$ is a random variable that is $\sigma(Y)$-measurable.For a random variable to be measurable with respect to a specific $\sigma$-algebra, its value cannot change within the finest non-empty sets (atoms) of that $\sigma$-algebra.

The atoms of $\sigma(Y)$ are exactly $A_0 = \{\omega_1, \omega_2\}$ and $A_1 = \{\omega_3, \omega_4\}$.If $E[X \mid Y]$ took different values for $\omega_1$ and $\omega_2$, we could construct a preimage mapping to a single value that isolates $\omega_1$ from $\omega_2$.

This preimage would not belong to $\sigma(Y)$, violating measurability. Therefore, $E[X \mid Y]$ must be constant on $\{\omega_1, \omega_2\}$ and constant on $\{\omega_3, \omega_4\}$.


---

# 4) Explicit Computation of $E[X \mid Y]$


Since $E[X \mid Y]$ is constant on the atoms of $\sigma(Y)$, its values are given by the conditional expectations over those events:

On the set $\{Y = 0\} = \{\omega_1, \omega_2\}$:$$E[X \mid Y = 0] = \frac{E[X \cdot \mathbb{1}_{\{Y=0\}}]}{P(Y=0)} = \frac{X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\})}{P(\{\omega_1\}) + P(\{\omega_2\})}$$$$E[X \mid Y = 0] = \frac{0(0.50) + 1(0.10)}{0.50 + 0.10} = \frac{0.10}{0.60} = \frac{1}{6}
$$On the set $\{Y = 1\} = \{\omega_3, \omega_4\}$:$$E[X \mid Y = 1] = \frac{E[X \cdot \mathbb{1}_{\{Y=1\}}]}{P(Y=1)} = \frac{X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\})}{P(\{\omega_3\}) + P(\{\omega_4\})}$$$$E[X \mid Y = 1] = \frac{0(0.20) + 1(0.20)}{0.20 + 0.20} = \frac{0.20}{0.40} = \frac{1}{2}$$Combining these using indicator functions gives:$$E[X \mid Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}$$

---


# 5) Verifying the Defining Property

---

### **Case 1: $A = \emptyset$**
Both sides trivially equal $0$.

---

### **Case 2: $A = \{\omega_1, \omega_2\}$**
$$\int_A E[X \mid Y] \, dP = \frac{1}{6} P(\{\omega_1\}) + \frac{1}{6} P(\{\omega_2\}) = \frac{1}{6}(0.50 + 0.10) = 0.10$$

$$\int_A X \, dP = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) = 0(0.50) + 1(0.10) = 0.10$$

> **Status:** Matches.

---

### **Case 3: $A = \{\omega_3, \omega_4\}$**
$$\int_A E[X \mid Y] \, dP = \frac{1}{2} P(\{\omega_3\}) + \frac{1}{2} P(\{\omega_4\}) = \frac{1}{2}(0.20 + 0.20) = 0.20$$

$$\int_A X \, dP = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) = 0(0.20) + 1(0.20) = 0.20$$

> **Status:** Matches.

---

### **Case 4: $A = \Omega$**
$$\int_\Omega E[X \mid Y] \, dP = 0.10 + 0.20 = 0.30$$

$$\int_\Omega X \, dP = 0.10 + 0.20 = 0.30$$

> **Status:** Matches.
