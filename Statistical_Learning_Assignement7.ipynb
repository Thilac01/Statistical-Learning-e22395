# Question 1
To show that $x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$, we derive its mean and covariance from the linear state transition model:

$$x^-_k = A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}$$

### 1. Mean Vector ($m^-_k$)

Taking the mathematical expectation $\mathbb{E}[\cdot]$ on both sides:

$$m^-_k = \mathbb{E}[x^-_k] = \mathbb{E}[A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}]$$

Using the linearity of the expectation operator:

$$m^-_k = A_{k-1}\mathbb{E}[x^+_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}]$$

Given that $\mathbb{E}[x^+_{k-1}] = m_{k-1}$ and the process noise is zero-mean ($\mathbb{E}[w_{k-1}] = 0$):

$$m^-_k = A_{k-1}m_{k-1} + G_{k-1}(0)$$

$$m^-_k = A_{k-1}m_{k-1}$$

---

### 2. Covariance Matrix ($P^-_k$)

The covariance matrix is defined as:

$$P^-_k = \mathbb{E}\left[ (x^-_k - m^-_k)(x^-_k - m^-_k)^T \right]$$

First, compute the deviation from the mean $(x^-_k - m^-_k)$:

$$x^-_k - m^-_k = (A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}) - A_{k-1}m_{k-1}$$

$$x^-_k - m^-_k = A_{k-1}(x^+_{k-1} - m_{k-1}) + G_{k-1}w_{k-1}$$

Substituting this back into the covariance formula yields:

$$P^-_k = \mathbb{E}\left[ \left( A_{k-1}(x^+_{k-1} - m_{k-1}) + G_{k-1}w_{k-1} \right) \left( A_{k-1}(x^+_{k-1} - m_{k-1}) + G_{k-1}w_{k-1} \right)^T \right]$$

Expanding the terms using the transpose property $(X+Y)^T = X^T + Y^T$:

$$P^-_k = \mathbb{E}\left[ A_{k-1}(x^+_{k-1} - m_{k-1})(x^+_{k-1} - m_{k-1})^TA_{k-1}^T + G_{k-1}w_{k-1}w_{k-1}^TG_{k-1}^T + \text{Cross Terms} \right]$$

Where the cross-covariance terms are:

$$\text{Cross Terms} = A_{k-1}(x^+_{k-1} - m_{k-1})w_{k-1}^TG_{k-1}^T + G_{k-1}w_{k-1}(x^+_{k-1} - m_{k-1})^TA_{k-1}^T$$

Applying the expectation operator individually to each term:

1. **State Term:** $\mathbb{E}\left[ A_{k-1}(x^+_{k-1} - m_{k-1})(x^+_{k-1} - m_{k-1})^TA_{k-1}^T \right] = A_{k-1}P_{k-1}A_{k-1}^T$
2. **Noise Term:** $\mathbb{E}\left[ G_{k-1}w_{k-1}w_{k-1}^TG_{k-1}^T \right] = G_{k-1}\Sigma_pG_{k-1}^T$
3. **Cross Terms:** Since the process noise $w_{k-1}$ is uncorrelated with the true state error, $\mathbb{E}\left[(x^+_{k-1} - m_{k-1})w_{k-1}^T\right] = 0$.

Combining these components:

$$P^-_k = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_pG_{k-1}^T$$



Because a linear combination of independent Gaussian distributions remains Gaussian, the prior state distribution is cleanly proved as:

$$x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$$

# Question 2

To show that the measurement vector $y_k$ follows the distribution $y_k \sim \mathcal{N}(H_k m^-_k, H_k P^-_k H_k^T + \Sigma_m)$, we derive its mean and covariance from the standard linear measurement model:

$$y_k = H_k x^-_k + v_k$$

Where $x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$ and the measurement noise is $v_k \sim \mathcal{N}(0, \Sigma_m)$.

### 1. Mean Vector ($\mathbb{E}[y_k]$)

Taking the mathematical expectation $\mathbb{E}[\cdot]$ on both sides:

$$\mathbb{E}[y_k] = \mathbb{E}[H_k x^-_k + v_k]$$

Using the linearity of the expectation operator:

$$\mathbb{E}[y_k] = H_k \mathbb{E}[x^-_k] + \mathbb{E}[v_k]$$

Substituting the known expectations $\mathbb{E}[x^-_k] = m^-_k$ and $\mathbb{E}[v_k] = 0$:

$$\mathbb{E}[y_k] = H_k m^-_k + 0$$

$$\mathbb{E}[y_k] = H_k m^-_k$$

---

### 2. Covariance Matrix ($\text{Cov}(y_k)$)

The covariance matrix of the measurement vector is defined as:

$$\text{Cov}(y_k) = \mathbb{E}\left[ (y_k - \mathbb{E}[y_k])(y_k - \mathbb{E}[y_k])^T \right]$$

First, compute the deviation from the mean $(y_k - \mathbb{E}[y_k])$:

$$y_k - \mathbb{E}[y_k] = (H_k x^-_k + v_k) - H_k m^-_k$$

$$y_k - \mathbb{E}[y_k] = H_k(x^-_k - m^-_k) + v_k$$

Substituting this back into the covariance formula yields:

$$\text{Cov}(y_k) = \mathbb{E}\left[ \left( H_k(x^-_k - m^-_k) + v_k \right) \left( H_k(x^-_k - m^-_k) + v_k \right)^T \right]$$

Expanding the expression using the transpose property $(X+Y)^T = X^T + Y^T$:

$$\text{Cov}(y_k) = \mathbb{E}\left[ H_k(x^-_k - m^-_k)(x^-_k - m^-_k)^T H_k^T + v_k v_k^T + \text{Cross Terms} \right]$$

Where the cross-covariance terms are:

$$\text{Cross Terms} = H_k(x^-_k - m^-_k)v_k^T + v_k(x^-_k - m^-_k)^T H_k^T$$

Applying the expectation operator individually to each term:

1. **State Covariance Term:** $\mathbb{E}\left[ H_k(x^-_k - m^-_k)(x^-_k - m^-_k)^T H_k^T \right] = H_k P^-_k H_k^T$
2. **Measurement Noise Term:** $\mathbb{E}[v_k v_k^T] = \Sigma_m$
3. **Cross Terms:** Since the measurement noise $v_k$ at time step $k$ is statistically independent of the predicted state error $(x^-_k - m^-_k)$, the cross-expectations collapse to zero: $\mathbb{E}\left[(x^-_k - m^-_k)v_k^T\right] = 0$.

Combining these components:

$$\text{Cov}(y_k) = H_k P^-_k H_k^T + \Sigma_m$$



Because a linear transformation of jointly Gaussian random variables yields another Gaussian distribution, the measurement distribution is proven as:

$$y_k \sim \mathcal{N}(H_k m^-_k, H_k P^-_k H_k^T + \Sigma_m)$$

# Question 3

To show that the joint distribution of the predicted state $x^-_k$ and the measurement $y_k$ is jointly Gaussian, we determine the joint mean vector and joint covariance matrix.

We represent the joint vector as a linear system of the underlying state error and noise:

$$\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} = \begin{bmatrix} I \\ H_k \end{bmatrix} x^-_k + \begin{bmatrix} 0 \\ I \end{bmatrix} v_k$$

### 1. Joint Mean Vector

Taking the mathematical expectation $\mathbb{E}[\cdot]$ of the joint vector:

$$\mathbb{E}\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} = \begin{bmatrix} \mathbb{E}[x^-_k] \\ \mathbb{E}[y_k] \end{bmatrix}$$

From our previous derivations, we know that $\mathbb{E}[x^-_k] = m^-_k$ and $\mathbb{E}[y_k] = H_k m^-_k$. Substituting these yields:

$$\mathbb{E}\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} = \begin{bmatrix} m^-_k \\ H_k m^-_k \end{bmatrix}$$

---

### 2. Joint Covariance Matrix

The joint covariance matrix is structured into a block matrix containing the auto-covariances on the diagonal and the cross-covariances on the off-diagonal:

$$\text{Cov}\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} = \begin{bmatrix} \text{Cov}(x^-_k, x^-_k) & \text{Cov}(x^-_k, y_k) \\ \text{Cov}(y_k, x^-_k) & \text{Cov}(y_k, y_k) \end{bmatrix}$$

We already know the two diagonal blocks from previous derivations:
* $\text{Cov}(x^-_k, x^-_k) = P^-_k$
* $\text{Cov}(y_k, y_k) = H_k P^-_k H_k^T + \Sigma_m$

Now, we derive the cross-covariance block $\text{Cov}(x^-_k, y_k)$:

$$\text{Cov}(x^-_k, y_k) = \mathbb{E}\left[ (x^-_k - m^-_k)(y_k - \mathbb{E}[y_k])^T \right]$$

Substitute the deviation expression for the measurement $(y_k - \mathbb{E}[y_k]) = H_k(x^-_k - m^-_k) + v_k$:

$$\text{Cov}(x^-_k, y_k) = \mathbb{E}\left[ (x^-_k - m^-_k) \left( H_k(x^-_k - m^-_k) + v_k \right)^T \right]$$

Expanding using the transpose property $(X+Y)^T = X^T + Y^T$:

$$\text{Cov}(x^-_k, y_k) = \mathbb{E}\left[ (x^-_k - m^-_k)(x^-_k - m^-_k)^T H_k^T \right] + \mathbb{E}\left[ (x^-_k - m^-_k)v_k^T \right]$$

Applying the expectation operator:
1. **State Cross Term:** $\mathbb{E}\left[ (x^-_k - m^-_k)(x^-_k - m^-_k)^T \right] H_k^T = P^-_k H_k^T$
2. **Noise Cross Term:** Since the predicted state error $(x^-_k - m^-_k)$ depends only on past noise components and is completely uncorrelated with the current measurement noise $v_k$, this expectation is zero: $\mathbb{E}\left[ (x^-_k - m^-_k)v_k^T \right] = 0$.

Thus:
$$\text{Cov}(x^-_k, y_k) = P^-_k H_k^T$$

By the symmetric property of covariance matrices, the remaining off-diagonal block is:
$$\text{Cov}(y_k, x^-_k) = \text{Cov}(x^-_k, y_k)^T = (P^-_k H_k^T)^T = H_k P^-_k$$



Assembling the derived mean vector blocks and covariance matrix blocks yields the complete joint Gaussian distribution:

$$\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} m^-_k \\ H_k m^-_k \end{bmatrix}, \begin{bmatrix} P^-_k & P^-_k H_k^T \\ H_k P^-_k & H_k P^-_k H_k^T + \Sigma_m \end{bmatrix} \right)$$

# Question 04

To prove the update step of the Kalman filter, we use the properties of **Conditional Gaussian Distributions**.

From our previous proof, the joint distribution of the predicted state $x^-_k$ and measurement $y_k$ is given by:

$$\begin{bmatrix} x^-_k \\ y_k \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} m^-_k \\ H_k m^-_k \end{bmatrix}, \begin{bmatrix} P^-_k & P^-_k H_k^T \\ H_k P^-_k & H_k P^-_k H_k^T + \Sigma_m \end{bmatrix} \right)$$

### The Conditional Gaussian Identity

For any jointly Gaussian random variables partitioned as:

$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mu_x \\ \mu_y \end{bmatrix}, \begin{bmatrix} \Sigma_{xx} & \Sigma_{xy} \\ \Sigma_{yx} & \Sigma_{yy} \end{bmatrix} \right)$$

The conditional distribution $(X \mid Y = y_{\text{obs}})$ is also Gaussian, $\mathcal{N}(\mu_{x|y}, \Sigma_{x|y})$, defined by:

$$\mu_{x|y} = \mu_x + \Sigma_{xy} \Sigma_{yy}^{-1} (y_{\text{obs}} - \mu_y)$$
$$\Sigma_{x|y} = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$

---

### Mapping to the Filter Parameters

By matching our joint matrix blocks to the identity variables, we map:

* $\mu_x = m^-_k$
* $\mu_y = H_k m^-_k$
* $\Sigma_{xx} = P^-_k$
* $\Sigma_{xy} = P^-_k H_k^T$
* $\Sigma_{yx} = H_k P^-_k$
* $\Sigma_{yy} = H_k P^-_k H_k^T + \Sigma_m$

Let us define the **Kalman Gain** matrix $K_k$ as:
$$K_k = \Sigma_{xy} \Sigma_{yy}^{-1} = P^-_k H_k^T (H_k P^-_k H_k^T + \Sigma_m)^{-1}$$

---

### 1. Updated Mean ($m_k$)

Applying the conditional mean formula:

$$m_k = \mu_x + \Sigma_{xy} \Sigma_{yy}^{-1} (y^{\text{obs}}_k - \mu_y)$$

Substitute our mapped filter terms:

$$m_k = m^-_k + P^-_k H_k^T (H_k P^-_k H_k^T + \Sigma_m)^{-1} (y^{\text{obs}}_k - H_k m^-_k)$$

Using our definition of $K_k$:

$$m_k = m^-_k + K_k (y^{\text{obs}}_k - H_k m^-_k)$$

---

### 2. Updated Covariance ($P_k$)

Applying the conditional covariance formula:

$$P_k = \Sigma_{xx} - \Sigma_{xy} \Sigma_{yy}^{-1} \Sigma_{yx}$$

Substitute our mapped filter terms:

$$P_k = P^-_k - \underbrace{P^-_k H_k^T (H_k P^-_k H_k^T + \Sigma_m)^{-1}}_{K_k} (H_k P^-_k)$$

Factoring out the Kalman Gain $K_k$:

$$P_k = P^-_k - K_k H_k P^-_k$$
$$P_k = (I - K_k H_k)P^-_k$$



Since the conditional distribution of a joint Gaussian distribution remains Gaussian, the measurement update equation holds true:

$$x^+_k \triangleq (x^-_k \mid y_k = y^{\text{obs}}_k) \sim \mathcal{N}(m_k, P_k)$$

where:
$$K_k = P^-_k H_k^T (H_k P^-_k H_k^T + \Sigma_m)^{-1}$$
$$m_k = m^-_k + K_k (y^{\text{obs}}_k - H_k m^-_k)$$
$$P_k = (I - K_k H_k)P^-_k$$
