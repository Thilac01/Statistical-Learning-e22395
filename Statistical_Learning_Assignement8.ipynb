# Q1 Statistical Mechanics of the Quantum Oscillator

The canonical partition function $Z(T)$ for a discrete set of energy levels is defined as the sum of the Boltzmann weights over all possible microstates:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{1}{k_B T} E_j}$$

### **Substitution of Energy Eigenvalues**
The energy eigenvalues for a quantum harmonic oscillator are given by:

$$E_j = \hbar\omega\left(j + \frac{1}{2}\right)$$

Let $\beta = \frac{1}{k_B T}$. Substituting the energy eigenvalues into the partition function definition yields:

$$Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar\omega \left(j + \frac{1}{2}\right)}$$

Using the laws of exponents ($a^{m+n} = a^m \cdot a^n$), we can split the exponential term into a state-independent ground-state term and a state-dependent term:

$$e^{-\beta \hbar\omega \left(j + \frac{1}{2}\right)} = e^{-\frac{1}{2}\beta \hbar\omega} \cdot e^{-\beta \hbar\omega j}$$

Substituting this back into the summation gives:

$$Z(T) = \sum_{j=0}^{\infty} \left( e^{-\frac{1}{2}\beta \hbar\omega} \cdot e^{-\beta \hbar\omega j} \right)$$

Since the term $e^{-\frac{1}{2}\beta \hbar\omega}$ does not depend on the summation index $j$, it behaves as a constant relative to the sum and can be factored out:

$$Z(T) = e^{-\frac{1}{2}\beta \hbar\omega} \sum_{j=0}^{\infty} \left(e^{-\beta \hbar\omega}\right)^j$$

The summation inside the expression is an infinite geometric series of the form:

$$\sum_{j=0}^{\infty} x^j = 1 + x + x^2 + x^3 + \dots$$

where the common ratio $x$ is defined as:

$$x = e^{-\beta \hbar\omega}$$

#### **Proof of Convergence**
For an infinite geometric series to converge, the absolute value of the common ratio must satisfy $|x| < 1$.

Given that $\hbar, \omega, k_B,$ and $T$ are all strictly positive physical constants, the exponent $-\beta \hbar\omega < 0$. Therefore:

$$0 < e^{-\beta \hbar\omega} < 1 \implies |x| < 1$$

Since this condition is satisfied, the series is strictly **convergent**.

### **Summation Evaluation and Final Simplification**
The sum of a convergent infinite geometric series is evaluated using the standard formula:

$$\sum_{j=0}^{\infty} x^j = \frac{1}{1 - x}$$

Applying this formula with $x = e^{-\beta \hbar\omega}$:

$$\sum_{j=0}^{\infty} \left(e^{-\beta \hbar\omega}\right)^j = \frac{1}{1 - e^{-\beta \hbar\omega}}$$

Substituting this closed-form evaluation back into our main expression for $Z(T)$ yields:

$$Z(T) = e^{-\frac{1}{2}\beta \hbar\omega} \cdot \left( \frac{1}{1 - e^{-\beta \hbar\omega}} \right) = \frac{e^{-\frac{1}{2}\beta \hbar\omega}}{1 - e^{-\beta \hbar\omega}}$$

Replacing $\beta$ with its fundamental thermodynamic definition $\frac{1}{k_B T}$, we arrive at the final expression:

$$Z(T) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

## 2. Thermodynamic Potentials

### **Helmholtz Free Energy, $F(T)$**
The Helmholtz Free Energy is defined as:
$$F(T) = -k_B T \ln Z$$

Let's compute $\ln Z$ first:
$$\ln Z = \ln\left( \frac{e^{-\frac{1}{2}\beta \hbar\omega}}{1 - e^{-\beta \hbar\omega}} \right) = \ln\left(e^{-\frac{1}{2}\beta \hbar\omega}\right) - \ln\left(1 - e^{-\beta \hbar\omega}\right)$$
$$\ln Z = -\frac{1}{2}\beta \hbar\omega - \ln\left(1 - e^{-\beta \hbar\omega}\right)$$

Now substitute this back into the formula for $F(T)$, remembering that $\beta k_B T = 1$:
$$F(T) = -k_B T \left[ -\frac{1}{2}\beta \hbar\omega - \ln\left(1 - e^{-\beta \hbar\omega}\right) \right]$$
$$F(T) = \frac{1}{2}\hbar\omega + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

---

### **Internal Energy, $U(T)$**
The Internal Energy is related to the partition function via:
$$U(T) = -\frac{\partial \ln Z}{\partial \beta}$$

Using our expression for $\ln Z$:
$$U(T) = -\frac{\partial}{\partial \beta} \left[ -\frac{1}{2}\beta \hbar\omega - \ln\left(1 - e^{-\beta \hbar\omega}\right) \right]$$
$$U(T) = \frac{1}{2}\hbar\omega + \frac{\partial}{\partial \beta} \left[ \ln\left(1 - e^{-\beta \hbar\omega}\right) \right]$$

Applying the chain rule to the second term:
$$\frac{\partial}{\partial \beta} \left[ \ln\left(1 - e^{-\beta \hbar\omega}\right) \right] = \frac{1}{1 - e^{-\beta \hbar\omega}} \cdot \left( -e^{-\beta \hbar\omega} \cdot (-\hbar\omega) \right) = \frac{\hbar\omega e^{-\beta \hbar\omega}}{1 - e^{-\beta \hbar\omega}}$$

To simplify, multiply the numerator and denominator by $e^{\beta \hbar\omega}$:
$$\frac{\hbar\omega}{e^{\beta \hbar\omega} - 1}$$

Combining both parts gives the final internal energy formula:
$$U(T) = \frac{1}{2}\hbar\omega + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

---

## 3. Entropy

The Entropy $S(T)$ is found using the thermodynamic relation:
$$S = -\left(\frac{\partial F}{\partial T}\right)$$

Differentiating our expression for $F(T)$ with respect to $T$:
$$F(T) = \frac{1}{2}\hbar\omega + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

Using the product rule on the second term:
$$S = -\left[ 0 + k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + k_B T \cdot \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \cdot \left( -e^{-\frac{\hbar\omega}{k_B T}} \cdot \frac{\hbar\omega}{k_B T^2} \right) \right]$$
$$S = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \frac{e^{-\frac{\hbar\omega}{k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

Multiplying the second term's fraction by $e^{\frac{\hbar\omega}{k_B T}}$ yields the explicit formula for Entropy:
$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T \left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)}$$

### **Verification of $U = F + TS$**
Let's compute $F + TS$:
$$F + TS = \left[ \frac{1}{2}\hbar\omega + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T \left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)} \right]$$

Distribute $T$ into the second bracket:
$$F + TS = \frac{1}{2}\hbar\omega + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

The logarithmic terms cancel out perfectly:
$$F + TS = \frac{1}{2}\hbar\omega + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1} = U(T)$$

$$\text{Verification Successful }$$

## 4. Heat Capacity and Hyperbolic Form

The constant-volume heat capacity is given by:
$$C_V = \left(\frac{\partial U}{\partial T}\right)$$

Differentiating $U(T)$ with respect to $T$:
$$U(T) = \frac{1}{2}\hbar\omega + \hbar\omega \left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-1}$$
$$C_V = \hbar\omega \cdot (-1)\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^{-2} \cdot \left(e^{\frac{\hbar\omega}{k_B T}} \cdot \left(-\frac{\hbar\omega}{k_B T^2}\right)\right)$$
$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2}$$

### **Proof of the Hyperbolic Form**
We want to show that this matches the target expression. Let's manipulate the fraction component by dividing the numerator and denominator by $e^{\frac{\hbar\omega}{k_B T}}$:
$$\frac{e^{\frac{\hbar\omega}{k_B T}}}{\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} = \frac{1}{e^{-\frac{\hbar\omega}{k_B T}}\left(e^{\frac{\hbar\omega}{k_B T}} - 1\right)^2} = \frac{1}{\left(e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}}\right)^2}$$

Using the identity $\sinh(x) = \frac{e^x - e^{-x}}{2} \implies e^x - e^{-x} = 2\sinh(x)$, let $x = \frac{\hbar\omega}{2k_B T}$:
$$\left(e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}}\right)^2 = \left(2\sinh\left(\frac{\hbar\omega}{2k_B T}\right)\right)^2 = 4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)$$

Substituting this back into our $C_V$ equation:
$$C_V = k_B \left(\frac{\hbar\omega}{k_B T}\right)^2 \frac{1}{4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

Bring the factor of 4 inside the squared term:
$$C_V = k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

$$\tag*{$\blacksquare$}$$

## 5. Asymptotic Behavior (Discussion)

In the high-temperature limit, the thermal energy is much greater than the quantum energy spacing ($k_B T \gg \hbar\omega$). This means the dimensionless parameter $x = \frac{\hbar\omega}{2k_B T} \to 0$.

Using the small-angle Taylor expansion for the hyperbolic sine function, $\sinh(x) \approx x$ when $x \ll 1$:
$$\sinh\left(\frac{\hbar\omega}{2k_B T}\right) \approx \frac{\hbar\omega}{2k_B T}$$

Plugging this approximation back into our compact $C_V$ formula:
$$C_V \approx k_B \left(\frac{\hbar\omega}{2k_B T}\right)^2 \frac{1}{\left(\frac{\hbar\omega}{2k_B T}\right)^2} = k_B$$



# The Maximum Entropy Distribution under Mean and Variance Constraints

We want to find a continuous probability density function (PDF) $f(x)$ for a random variable $X \in (-\infty, \infty)$ that maximizes the differential entropy $H(f)$, subject to three physical constraints: normalization, fixed mean, and fixed variance.

### **1. Setting up the Optimization Problem**

The **differential entropy** to maximize is:
$$H(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx$$

Our optimization constraints are:
1. **Normalization:** $\int_{-\infty}^{\infty} f(x) \, dx = 1$
2. **Mean Constraint:** $\int_{-\infty}^{\infty} x f(x) \, dx = \mu$
3. **Variance Constraint:** $\int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx = \sigma^2$

---

### **2. The Lagrange Multipliers Method**

We define the Lagrangian function $L(f)$ by introducing three constants (Lagrange multipliers) $\lambda_0$, $\lambda_1$, and $\lambda_2$:

$$L(f) = -\int_{-\infty}^{\infty} f(x) \ln f(x) \, dx - \lambda_0 \left( \int_{-\infty}^{\infty} f(x) \, dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x f(x) \, dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 f(x) \, dx - \sigma^2 \right)$$

Combining everything under a single integral gives:
$$L(f) = \int_{-\infty}^{\infty} \left[ -f(x)\ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x-\mu)^2 f(x) \right] dx + \lambda_0 + \lambda_1\mu + \lambda_2\sigma^2$$

To maximize this functional with respect to $f(x)$, we take the functional derivative inside the integrand with respect to $f(x)$ and set it to zero:

$$\frac{\partial}{\partial f} \left[ -f(x)\ln f(x) - \lambda_0 f(x) - \lambda_1 x f(x) - \lambda_2 (x-\mu)^2 f(x) \right] = 0$$

$$-(\ln f(x) + 1) - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0$$

Solving for $\ln f(x)$:
$$\ln f(x) = -(1 + \lambda_0) - \lambda_1 x - \lambda_2 (x-\mu)^2$$

Exponentials of both sides yield the general functional form of our target distribution:
$$f(x) = e^{-(1 + \lambda_0)} e^{-\lambda_1 x} e^{-\lambda_2 (x-\mu)^2}$$

---

### **3. Solving for the Constants**

To match the traditional Gaussian form symmetrically around $\mu$, we can look ahead at our linear parameter $\lambda_1$. Because the domain is infinite $(-\infty, \infty)$ and symmetric around $\mu$, any linear shift parameter $\lambda_1$ outside the squared variation term would break the expected centering on the mean $\mu$, meaning $\lambda_1 = 0$.

Letting $C = e^{-(1 + \lambda_0)}$, our simplified expression becomes:
$$f(x) = C e^{-\lambda_2 (x-\mu)^2}$$

Now, we solve for $C$ and $\lambda_2$ using our normalization and variance properties.

#### **Finding $C$ (Normalization)**
Using the standard Gaussian integral $\int_{-\infty}^{\infty} e^{-a(x-b)^2} \, dx = \sqrt{\frac{\pi}{a}}$:
$$\int_{-\infty}^{\infty} C e^{-\lambda_2 (x-\mu)^2} \, dx = 1 \implies C \sqrt{\frac{\pi}{\lambda_2}} = 1 \implies C = \sqrt{\frac{\lambda_2}{\pi}}$$

#### **Finding $\lambda_2$ (Variance)**
Now we substitute $f(x) = \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x-\mu)^2}$ into our variance constraint equation:
$$\int_{-\infty}^{\infty} (x-\mu)^2 \left( \sqrt{\frac{\lambda_2}{\pi}} e^{-\lambda_2 (x-\mu)^2} \right) \, dx = \sigma^2$$

Using the known standard integral evaluation $\int_{-\infty}^{\infty} y^2 e^{-a y^2} \, dy = \frac{1}{2a}\sqrt{\frac{\pi}{a}}$ (where $y = x-\mu$ and $a = \lambda_2$):
$$\sqrt{\frac{\lambda_2}{\pi}} \cdot \left( \frac{1}{2\lambda_2} \sqrt{\frac{\pi}{\lambda_2}} \right) = \sigma^2$$

$$\frac{1}{2\lambda_2} = \sigma^2 \implies \lambda_2 = \frac{1}{2\sigma^2}$$

---

### **4. Final Formulation**

Now we substitute our solved parameter values back into the density function expressions:
* $\lambda_2 = \frac{1}{2\sigma^2}$
* $C = \sqrt{\frac{\lambda_2}{\pi}} = \sqrt{\frac{1}{2\pi\sigma^2}} = \frac{1}{\sqrt{2\pi\sigma^2}}$

Plugging $C$ and $\lambda_2$ back directly into $f(x)$:

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)$$



# Q. Multiclass Logistic Regression: Automatic Classification of Emails

## 2. Setting Up the Probabilistic System

We need to resolve the conditional distribution $P(Y=1 \mid \mathbf{x})$ for each scenario to construct or train the logistic regression model. Let's define the conditional probabilities as parameters to be solved:
$$p_A = P(Y=1 \mid \mathbf{x}_A)$$
$$p_B = P(Y=1 \mid \mathbf{x}_B)$$
$$p_C = P(Y=1 \mid \mathbf{x}_C)$$
$$p_D = P(Y=1 \mid \mathbf{x}_D)$$
$$p_E = P(Y=1 \mid \mathbf{x}_E)$$

Using the law of total probability, the joint probability can be split into $P(x_i=1, Y=1) = \sum_{\mathbf{x}} x_i P(Y=1 \mid \mathbf{x}) P(\mathbf{x})$. Let's apply this expansion to our three constraints:

### **Expansion of Constraint 1 ($x_1 = 1$):**
Feature $x_1$ is active only in **Scenario A** $(1,0,0)^T$ and **Scenario E** $(1,1,0)^T$.
$$P(\mathbf{x}_A) \cdot p_A + P(\mathbf{x}_E) \cdot p_E = 0.05$$
$$0.40 p_A + 0.05 p_E = 0.05 \quad \implies \quad 8p_A + p_E = 1 \quad \text{--- (Eq. 1)}$$

### **Expansion of Constraint 2 ($x_2 = 1$):**
Feature $x_2$ is active in **Scenario B** $(0,1,0)^T$, **Scenario D** $(0,1,1)^T$, and **Scenario E** $(1,1,0)^T$.
$$P(\mathbf{x}_B) \cdot p_B + P(\mathbf{x}_D) \cdot p_D + P(\mathbf{x}_E) \cdot p_E = 0.15$$
$$0.25 p_B + 0.15 p_D + 0.05 p_E = 0.15 \quad \implies \quad 5p_B + 3p_D + p_E = 3 \quad \text{--- (Eq. 2)}$$

### **Expansion of Constraint 3 ($x_3 = 1$):**
Feature $x_3$ is active only in **Scenario C** $(0,0,1)^T$ and **Scenario D** $(0,1,1)^T$.
$$P(\mathbf{x}_C) \cdot p_C + P(\mathbf{x}_D) \cdot p_D = 0.16$$
$$0.05 p_C + 0.15 p_D = 0.16 \quad \implies \quad p_C + 3p_D = 3.2 \quad \text{--- (Eq. 3)}$$

---

## 3. Resolving the Probabilities with Maximum Entropy

Since we have 5 operational parameters and only 3 constraint equations, the system is underdetermined. In information theory and machine learning (logistic regression), we choose the distribution that satisfies the constraints while maximizing entropy (introducing no extra assumptions).

However, we must also strictly respect the probability boundaries $0 \le p_i \le 1$. Let's examine **Equation 3**:
$$p_C + 3p_D = 3.2$$

Because $p_C \le 1$, the minimum possible value for $3p_D$ happens when $p_C$ is maximized ($p_C = 1$):
$$1 + 3p_D = 3.2 \implies 3p_D = 2.2 \implies p_D = \frac{2.2}{3} \approx 0.733$$

Conversely, because $p_D \le 1$, the maximum value $3p_D$ can take is $3(1) = 3$. Substituting $p_D = 1$ into Eq. 3 yields:
$$p_C + 3(1) = 3.2 \implies p_C = 0.2$$

Thus, physical limits constrain these scenarios to the ranges:
* $0.733 \le p_D \le 1.0$
* $0.20 \le p_C \le 1.0$

If we optimize the logistic regression formulation directly matching these constraints via maximum entropy, the unique parametric solutions for the conditional routing probabilities yield:

## 4. Logistic Regression Parameter Formulation

In standard binary logistic regression, the probability model is structured via the sigmoid function:
$$P(Y=1 \mid \mathbf{x}) = \sigma(\mathbf{w}^T \mathbf{x} + b) = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}$$

Where the log-odds (logit) is a linear combination of the features:
$$\ln\left( \frac{P(Y=1 \mid \mathbf{x})}{1 - P(Y=1 \mid \mathbf{x})} \right) = w_1 x_1 + w_2 x_2 + w_3 x_3 + b$$

    $$w_2 + w_3 + b = \ln\left(\frac{0.80}{1-0.80}\right) = \ln(4) \approx 1.386$$

### **Solving for parameters:**

# Q3. Asset Pricing and Portfolio Optimization

We want to find a continuous probability density function $p(S_T)$ that maximizes the continuous Shannon entropy:

$$H(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T$$

We have to meet four strict conditions (one fundamental probability rule + our three market observations):

1. **Normalization (The absolute rule of probability):** $$\int_{0}^{\infty} p(S_T) \, dS_T = 1$$
2. **Risk-Free Growth (Expected Value):** $$\int_{0}^{\infty} S_T \cdot p(S_T) \, dS_T = 102$$
3. **Market Volatility (Expected Second Moment):** $$\int_{0}^{\infty} S_T^2 \cdot p(S_T) \, dS_T = 10,600$$
4. **Asymmetric Market Sentiment (Logarithmic Skew):** $$\int_{0}^{\infty} \ln(S_T) \cdot p(S_T) \, dS_T = 4.60$$

---


To optimize our entropy equation while strictly obeying all four rules, we bundle everything together into a single master equation using **Lagrange Multipliers** ($\lambda_0, \lambda_1, \lambda_2, \lambda_3$):

$$L(p) = -\int_{0}^{\infty} p(S_T) \ln p(S_T) \, dS_T - \lambda_0 \left( \int_{0}^{\infty} p(S_T) \, dS_T - 1 \right) - \lambda_1 \left( \int_{0}^{\infty} S_T \, p(S_T) \, dS_T - 102 \right) - \lambda_2 \left( \int_{0}^{\infty} S_T^2 \, p(S_T) \, dS_T - 10,600 \right) - \lambda_3 \left( \int_{0}^{\infty} \ln(S_T) \, p(S_T) \, dS_T - 4.60 \right)$$

We group all the parts living inside the integrals under a single integral block:

$$L(p) = \int_{0}^{\infty} \left[ -p(S_T)\ln p(S_T) - \lambda_0 p(S_T) - \lambda_1 S_T p(S_T) - \lambda_2 S_T^2 p(S_T) - \lambda_3 \ln(S_T) p(S_T) \right] dS_T + \text{constant multiplier terms}$$

---



To find the absolute peak value for our probability distribution function, we take the functional derivative of the inside bracket with respect to $p(S_T)$ and set it to zero:

$$\frac{\partial}{\partial p} \left[ -p(S_T)\ln p(S_T) - \lambda_0 p(S_T) - \lambda_1 S_T p(S_T) - \lambda_2 S_T^2 p(S_T) - \lambda_3 \ln(S_T) p(S_T) \right] = 0$$

$$- \ln p(S_T) - 1 - \lambda_0 - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

Now, let's isolate the term $\ln p(S_T)$:

$$\ln p(S_T) = -(1 + \lambda_0) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)$$

To completely free $p(S_T)$, we take the exponential ($e^x$) of both sides:

$$p(S_T) = e^{-(1 + \lambda_0)} \cdot e^{-\lambda_1 S_T} \cdot e^{-\lambda_2 S_T^2} \cdot e^{-\lambda_3 \ln(S_T)}$$

---


We can simplify the look of this function dramatically with two basic algebra steps:
* Let the constant scaling term be $C = e^{-(1 + \lambda_0)}$.
* Use exponential rules to rewrite the last term: $e^{-\lambda_3 \ln(S_T)} = e^{\ln(S_T^{-\lambda_3})} = S_T^{-\lambda_3}$.

Plugging these back in gives us the exact functional family of our distribution:

$$p(S_T) = C \cdot S_T^{-\lambda_3} \cdot e^{-\lambda_1 S_T - \lambda_2 S_T^2}$$

> **What type of distribution is this?** > This belongs to the **Generalized Gamma Distribution** family. The quadratic term $e^{-\lambda_2 S_T^2}$ actively suppresses hyper-extreme values to account for volatility limits, while the power term $S_T^{-\lambda_3}$ allows for the heavy, asymmetric downside "fat tail" that standard models routinely miss.

---



Because the exponents include combinations of linear, quadratic, and logarithmic variations, calculating the exact numerical values for $C, \lambda_1, \lambda_2,$ and $\lambda_3$ analytically (by hand) is impossible. Instead, quantitative managers solve for them using a numerical root-finding optimizer (like the Newton-Raphson method or Scipy's `minimize` function).

The parameters must be numerically solved to satisfy the system of target moments simultaneously:

$$\begin{cases}
\int_{0}^{\infty} C S_T^{-\lambda_3} e^{-\lambda_1 S_T - \lambda_2 S_T^2} \, dS_T = 1 \\
\int_{0}^{\infty} S_T \cdot C S_T^{-\lambda_3} e^{-\lambda_1 S_T - \lambda_2 S_T^2} \, dS_T = 102 \\
\int_{0}^{\infty} S_T^2 \cdot C S_T^{-\lambda_3} e^{-\lambda_1 S_T - \lambda_2 S_T^2} \, dS_T = 10,600 \\
\int_{0}^{\infty} \ln(S_T) \cdot C S_T^{-\lambda_3} e^{-\lambda_1 S_T - \lambda_2 S_T^2} \, dS_T = 4.60
\end{cases}$$

Once these four numerical parameters are calculated via software, the explicit, least-biased custom probability curve is complete.
